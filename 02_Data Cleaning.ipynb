```python
import pandas as pd 

df = pd.read_csv("jobs_new.csv")
```

```python
df.head()
```

```python
# Information about table 

df.info()
```

```python
# Checking how many rows and columns 

df.shape
```

```python
# Checking columns in table or data

df.columns
```

```python
# Checking duplicates records

df.duplicated().sum()
```

```python
# Remove duplicates data

df =df.drop_duplicates()
```

```python
# Check remain any duplicates data

print(df.shape)
print(df.duplicated().sum())
```

```python
# Checking null values 

df.isnull().sum()
```

```python
# Adding "unknown" in company missing values 
 
df["Company"] = df["Company"].fillna("Unknown")
```

```python
# Adding "Not Specified" in Contract_type and Contract_time 

df["Contract_Type"] = df["Contract_Type"].fillna("Not Specified")

df["Contract_Time"] = df["Contract_Time"].fillna("Not Specified")
```

```python
# Check data types

df.dtypes
```

```python
# Removes Extra Space

df["Title"] = df["Title"].str.strip()
df["Company"] = df["Company"].str.strip()
df["Location"] = df["Location"].str.strip()
df["Contract_Type"] = df["Contract_Type"].str.strip()
df["Contract_Time"] = df["Contract_Time"].str.strip()
df["Category"] = df["Category"].str.strip()
df["City_Searched"] = df["City_Searched"].str.strip()
df["Role_Searched"] = df["Role_Searched"].str.strip()
```

```python
# Checking Unique values in company 

print(df["Company"].unique())
print(df["Role_Searched"].unique())
```

```python
# Standardise Text

df["Title"] = df["Title"].str.title()
df["Company"] = df["Company"].str.title()
df["Location"] = df["Location"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Contract_Time"] = df["Contract_Time"].str.title()
df["Category"] = df["Category"].str.title()
df["City_Searched"] = df["City_Searched"].str.title()
df["Role_Searched"] = df["Role_Searched"].str.title()
```

```python
# Check Outliers(salary)

df.describe()
```

```python
# Save Cleaned Datasets

df.to_csv("jobs_cleaned.csv", index = False)
```
